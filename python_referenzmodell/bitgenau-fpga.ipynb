### Leaky Integrate and Fire (DELTA SYNAPSE): 
*Refernzmodell mit den selben Bitbreiten wie System Verilog Code*
Identisch mit dem tb_lif_delta.tb

```python
# Leaky Integrate and Fire Modul, bitgenauer Vergleich
# Q2.14 Format
v_m = 0          # Ruhepotential = 0
v_th = 16384 # 1.0 in signed 16 bit

decay_shift_1 = 4 #entspricht >>> 4 = / 2 **4
decay_shift_2 = 5 # >>> 5 bzw / 2**5

# wenn 16 bit inklusive Vorzeichen
# Wertebereich: von -2^15 bis 2^15-1
MIN = -32768
MAX = 32767

def zerfall_mem(v_m, decay1, decay2):
    v_m = v_m - (v_m >> decay1) - (v_m >> decay2)
    return v_m

def check_overflow(v_decayed):
    if v_decayed >= MAX:
        return MAX
    elif v_decayed <= MIN:
        return MIN
    else: return v_decayed

def update (v_m, input_current):
    v_m += input_current
    return v_m

def check_threshold(v_m):
    if v_m >= v_th:
        v_m = 0
    else:
        pass
    return v_m

input_currents = [576, 0, 32600, -600, -32600]
sim_dauer = len(input_currents)

for i in range (sim_dauer):
  v_m = zerfall_mem(v_m, decay_shift_1, decay_shift_2)
  print(f'Nach dem Zerfall: {v_m}')
  v_m = update(v_m, input_currents[i])
  print(f'Nach dem Update: {v_m}')
  v_m = check_overflow(v_m)
  print(f'Nach dem Overflow Check: {v_m}')
  v_m = check_threshold(v_m)
  print(f'Nach dem Threshold: {v_m} \n')
```
